```python
# 1. Ternary Conditional Operator (One-liner if)

age = 18
status = "Adult" if age >= 18 else "Minor"
print(status)
```

```python
# 2. Multiple Conditions (Logical Operators) {and, or, not}

x = 15
if x > 10 and x < 20:
    print("x is between 10 and 20")
```

```python
# 3. Nested Conditionals

x = 5
if x > 0:
    if x < 10:
        print("x is a positive single-digit number")
```

```python
# 4. Using in and not in

letter = 'a'
if letter in ['a', 'e', 'i', 'o', 'u']:
    print("Vowel")

letter = 'b'
if letter not in ['a', 'e', 'i', 'o', 'u']:
    print("Consonant")
```

```python
# 5. Match-Case (Python 3.10+)

day = "Monday"

match day:
    case "Monday":
        print("Start of the week")
    case "Friday":
        print("Almost weekend")
    case _:
        print("Just another day")
```

```python
# 6. Chained Comparisons
x = 7
if 5 < x < 10:
    print("x is between 5 and 10")
```

```python
# 7. Using Boolean Expressions
user_logged_in = True
admin = False

if user_logged_in and not admin:
    print("Welcome, user!")
```
