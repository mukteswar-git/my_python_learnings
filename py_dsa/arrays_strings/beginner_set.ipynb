```python
# Print Alternates
# Given an array of integers, print every alternate element starting from index 0.

def print_alternate_elements(arr):
    for i in range(0, len(arr), 2): # Step of 2
        print(arr[i], end=' ')

# Test case 1
print("Output for [1, 2, 3, 4, 5, 6]:")
print_alternate_elements([1, 2, 3, 4, 5, 6])

# Test case 2
print("\nOutput for [10, 20, 30]:")
print_alternate_elements([10, 20, 30])
```

```python
# Reverse an Array
# Given an array, reverse the elements in-place and print the result.
arr = [1, 2, 3, 4, 5]
arr.reverse()
print(arr)

def reverse_array_builtin(arr):
    reversed_arr = arr[::-1]
    for ele in reversed_arr:
        print(ele, end=' ')

print("\nOutput for [1, 2, 3, 4, 5]:")
reverse_array_builtin([1, 2, 3, 4, 5])

print("\nOutput for [10, 20]:")
reverse_array_builtin([10, 20])
```

```python
# Linear Search
# Linear search is the simplest search algorithm. It checks every element in the list one by one until it finds the target value or reaches the end.

def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i # Found the target, return index
    return -1 # Return -1 if not found
        
arr = [10, 20, 300, 40, 50]
target = 30
result = linear_search(arr, target)

if result != -1:
    print(f"Element found at index {result}")
else:
    print("Element not found")
```
